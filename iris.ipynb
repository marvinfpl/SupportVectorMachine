```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import hinge_loss, classification_report
import importlib
import kernelSoftMarginSvm
importlib.reload(kernelSoftMarginSvm)
from kernelSoftMarginSvm import KernelSoftMarginSvmClassifier
import numpy as np

iris = datasets.load_iris()
X = iris.data
y = iris.target
mask = y != 2
X = X[mask]
y = y[mask]
y[y == 0] = -1
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=True)

print(X_test.shape)
model = KernelSoftMarginSvmClassifier(d=3, kernel='polynomial')
model.fit(np.array(X_train), np.array(y_train))
print(f'alpha: {model.alpha}\n'
      f'support vectors: {model.support_vectors}\n'
      f'support labels: {model.support_labels} \n')
```

```python
def sigmoid(x: np.ndarray):
    return 1 / (1 + np.exp(-x))

pred = model.predict(np.array(X_test))
report = classification_report(y_test, pred[0], target_names=['-1', '+1'])
loss = hinge_loss(y_test, pred[1])
print(f'Log loss: {loss} \n')
print(report)
```
